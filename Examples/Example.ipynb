```python
import numpy as np

# Probabilidades a priori para moeda justa (H1) e moeda enviesada (H2)
p_H1 = 0.5  # Prior para H1: moeda justa
p_H2 = 0.5  # Prior para H2: moeda enviesada (P(heads)=0.7)

# Likelihood do dado (resultado do lançamento) sob cada hipótese
def likelihood(data, hypothesis):
    if hypothesis == "H1":   # moeda justa
        return 0.5
    elif hypothesis == "H2": # moeda enviesada
        return 0.7 if data == 1 else 0.3

# Atualização bayesiana
def bayesian_update(prior_H1, prior_H2, data):
    # Likelihoods
    likelihood_H1 = likelihood(data, "H1")
    likelihood_H2 = likelihood(data, "H2")

    # Evidência total P(D)
    p_data = likelihood_H1 * prior_H1 + likelihood_H2 * prior_H2

    # Posteriores
    posterior_H1 = (likelihood_H1 * prior_H1) / p_data
    posterior_H2 = (likelihood_H2 * prior_H2) / p_data

    return posterior_H1, posterior_H2

# Gerar 20 lançamentos aleatórios: 1 = cara, 0 = coroa
coin_flips = np.random.choice([1, 0], size=20)

# Começamos com os priors
p_H1_posterior = p_H1
p_H2_posterior = p_H2

# Atualização após cada lançamento
for flip in coin_flips:
    p_H1_posterior, p_H2_posterior = bayesian_update(
        p_H1_posterior, p_H2_posterior, flip
    )

    print(
        f"Após observar {flip}: "
        f"P(H1 | Dados) = {p_H1_posterior:.4f}, "
        f"P(H2 | Dados) = {p_H2_posterior:.4f}"
    )
```
